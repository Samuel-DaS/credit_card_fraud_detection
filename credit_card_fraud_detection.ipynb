Data Visualization

```python
#Rarity of the fraudulent transactions
counts = credit_card_data['Class'].value_counts().sort_index() #selects Class column (0's and 1's)
#value.counts() counts how many 0s and 1s are in the dataset
#sort.index() sorts by class label (0 first, 1 second)
print(counts)                                                  

#Data Visualization
plt.figure(figsize=(5,4))
sns.barplot(x=counts.index, y=counts.values)
plt.xticks([0,1],['Normal (0)', 'Fraud (1)'])
plt.ylabel('Number of transactions')
plt.title('Class Balance: normal vs fraud')
plt.tight_layout()
plt.show()
```

```python
#Amound distribution overall
plt.figure(figsize=(5,4))
sns.histplot(credit_card_data['Amount'], bins=60, kde=False)
plt.xlabel('Transaction Amount')
plt.title('Distribution of transaction amounts (all)')
plt.tight_layout()
plt.show()

#Amount distribution by class
plt.figure(figsize=(5,4))
sns.histplot(
    data = credit_card_data,
    x = 'Amount',
    bins = 60,
    stat = 'density',
    common_norm = False
)
plt.xlabel('Transaction Amount')
plt.title('Amount distribution by class')
plt.tight_layout()
plt.show()
```

```python
#Amount boxplot by class
plt.figure(figsize=(5,4))
sns.boxplot(data=credit_card_data, x='Class', y='Amount')
plt.xticks([0,1], ['Normal (0)', 'Fraud (1)'])
plt.title('Amount by class (boxplot)')
plt.tight_layout()
plt.show()
```

```python
#Time Patterns
df = credit_card_data.copy()
df['Hour'] = (df['Time'] // 3600).astype(int)

#Counts per hour for fraud vs normal
hourly = df.groupby(['Hour', 'Class']).size().reset_index(name='count')
plt.figure(figsize=(8,4))
sns.lineplot(data=hourly, x='Hour', y='count', hue='Class')
plt.xticks(range(0,48,2))
plt.title('Transactions per hour by class')
plt.tight_layout()
plt.show()
```

```python
#Scatter plot
sample = credit_card_data.sample(10000, random_state=42)
plt.figure(figsize=(7,5))
sns.scatterplot(data=sample, x='V1', y='V2', hue='Class', alpha=0.6, s=20)
plt.title ('V1 vs V2 colored by class (sampled)')
plt.show()
```

```python
#Heatmap
corr = credit_card_data.sample(20000, random_state=42).corr(numeric_only=True)

plt.figure(figsize=(5,4))
sns.heatmap(corr, cmap='coolwarm', center=0, square=False)
plt.title('Feature correlation heatmap (sample)')
plt.tight_layout()
plt.show()
```
